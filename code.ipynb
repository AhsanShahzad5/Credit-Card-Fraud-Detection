```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
credit_card_data = pd.read_csv('creditcard.csv')
```

**This Dataset is highly unblanced**

0 --> Normal Transaction

1 --> fraudulent transaction

```python
# separating the data for analysis
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]

print(legit.shape)
print(fraud.shape)
```

```python
# statistical measures of the data
legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

```python
# compare the values for both transactions
credit_card_data.groupby('Class').mean()
```

### handling imbalance

here we use a very simpler technique , **under sampling** , we take only 492 legit ones , so we have a balanced distribution

```python
legit_sample = legit.sample(n=492)
```

```python
new_dataset = pd.concat([legit_sample, fraud], axis=0)
```

```python
new_dataset.head()

# DETAILS ABOUT TRANSACTIONS HAVE BEEN CONVERTED TO THESE NUMERICAL VALUES USING PCA 
# WE Have 2 days of transaction data
```

```python
new_dataset.shape # now we have 984 total data points
```

```python
new_dataset['Class'].value_counts()
```

**Splitting the data into Features & Targets**

```python
X = new_dataset.drop(columns='Class')
Y = new_dataset['Class']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

### Model Training

```python
model = LogisticRegression()
```

```python
# training the Logistic Regression Model with Training Data
model.fit(X_train, Y_train)
```

```python
# accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy on Training data : ', training_data_accuracy)
```

```python
# accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy score on Test Data : ', test_data_accuracy)
```

```python
from sklearn.metrics import classification_report

report = classification_report(Y_test, X_test_prediction)
print(report)
```

## Trying xgboost on it

```python
import xgboost as xgb

# model_xgb = xgb.XGBClassifier(
#     n_estimators=100, 
#     learning_rate=0.1, 
#     max_depth=3, 
#     random_state=42
# )


# above one xgb was causing slight ovetfitting , 99+ accuracy on train and 93 on test

model_xgb = xgb.XGBClassifier(
    n_estimators=100, 
    learning_rate=0.1, 
    max_depth=2, 
    subsample=0.8,         # Uses 80% of rows per tree to prevent memorization
    colsample_bytree=0.8,  # Uses 80% of features per tree
    reg_alpha=0.1,         # L1 regularization
    reg_lambda=1.0,        # L2 regularization
    random_state=42
)
```

```python
# 3. Fit the model on the training data
model_xgb.fit(X_train, Y_train)
```

```python
y_pred_train = model_xgb.predict(X_train)
```

```python
print("Accuracy of xgb on training data", accuracy_score(Y_train, y_pred_train))
print("\nClassification Report:\n", classification_report(Y_train, y_pred_train))
```

```python
# 4. Generate predictions on the test set
y_pred = model_xgb.predict(X_test)
```

```python
# 5. Evaluate the results
print("Test data Accuracy:", accuracy_score(Y_test, y_pred))
print("\nTest data  Classification Report:\n", classification_report(Y_test, y_pred))
```

There is very slight overfitting as we can see above , but its not a big issue , its completely normal

```python
from sklearn.model_selection import cross_val_score

# Check performance across 5 different splits 
cv_scores = cross_val_score(model_xgb, X_train, Y_train, cv=5, scoring='accuracy')

print("Cross-Validation Accuracies:", cv_scores)
print("Mean CV Accuracy:", cv_scores.mean())
```
